```python
import pandas as pd
import numpy as np
import os
from numpy import triu
from gensim.models import word2vec
```

```python
def load_csvs(path):
    # 디렉터리 경로 설정
    directory_path = path # 예: '/path/to/csv/CLTUR'

    file_list = os.listdir(directory_path)

    # CSV 파일들만 필터링하여 읽기
    csv_files = [file for file in file_list if file.endswith('.csv')]
    dfs = []
    for csv_file in csv_files:
        file_path = os.path.join(directory_path, csv_file)
        df = pd.read_csv(file_path)
        dfs.append(df)
    return dfs
```

```python
def preprocess_dup(df):
	df = df.drop_duplicates(subset=['DATA_TITLE_NM'])
	return df.reset_index()
```

```python
import re

def preprocessing(text):
    # 개행문자 제거
    text = re.sub('\\\\n', ' ', text)
    # # 한글, 영문만 남기고 모두 제거
    text = re.sub('[^가-힣ㄱ-ㅎㅏ-ㅣa-zA-Z]', ' ', text)
    return text
```

```python

final_df = pd.concat([merged_df_CLTUR, merged_df_LCFD, merged_df_OLD_LOAD, merged_df_MRKT])
final_df['sentence']=final_df.apply(lambda row: ' '.join(row.values.astype(str)), axis=1)

sentences=final_df['sentence'].apply(preprocessing)
sentences
```

```python
from soynlp.tokenizer import RegexTokenizer

tokenizer = RegexTokenizer()
tokens = sentences.apply(tokenizer.tokenize)
tokens[:]
```

```python
model = word2vec.Word2Vec(tokens, min_count=1)
```

```python
# 지역 이름과 키워드 열만 유지
final_df = final_df[['SIGNGU_NM', 'CORE_KWRD_CN']].reset_index().drop(columns=['index'])
final_df['CORE_KWRD_CN'] = final_df['CORE_KWRD_CN'].astype(str)
# 시군구 이름으로 키워드 행 묶기
final_df = final_df.groupby('SIGNGU_NM')['CORE_KWRD_CN'].apply(lambda x: ', '.join(x)).reset_index()
final_df
# print(final_df[final_df['SIGNGU_NM'] == '가평군']['CORE_KWRD_CN'])
```

```python
def preprocess_text(text):
    if isinstance(text, str):
        return re.sub(r'[^\w\s]', '', text)
    else:
        return ''
```

```python
def find_similar_region(input_region):
    input_keywords = preprocess_text(final_df[final_df['SIGNGU_NM'] == input_region]['CORE_KWRD_CN'].values[0])
    similar_regions = []
    for index, row in final_df.iterrows():
        region = row['SIGNGU_NM']
        if region != input_region:
            region_keywords = preprocess_text(row['CORE_KWRD_CN'])
            if (region_keywords):
              # print(input_keywords, region_keywords)
              similarity = model.wv.n_similarity(input_keywords, region_keywords)
              similar_regions.append((region, similarity))

    similar_regions.sort(key=lambda x: x[1], reverse=True)
    top_similar_regions = [region for region, _ in similar_regions[:5]]
    # 가장 낮은 평균 유사도 초기화
    lowest_avg_similarity = float('inf')
    lowest_avg_similarity_region = None

    # 각 상위 5개 지역에 대해 나머지 지역들과의 평균 유사도 계산
    for top_region in top_similar_regions:
        similarities = []
        for index, row in final_df.iterrows():
            region = row['SIGNGU_NM']
            if region not in top_similar_regions:
                region_keywords = preprocess_text(row['CORE_KWRD_CN'])
                similarity = model.wv.n_similarity(input_keywords, region_keywords)
                similarities.append(similarity)

        # 평균 유사도 계산
        avg_similarity = sum(similarities) / len(similarities)

        # 가장 낮은 평균 유사도와 그 지역 저장
        if avg_similarity < lowest_avg_similarity:
            lowest_avg_similarity = avg_similarity
            lowest_avg_similarity_region = top_region

    return similar_regions, lowest_avg_similarity_region, lowest_avg_similarity

# 테스트
input_region = '화성시'
similar_regions, lowest_region, lowest_similarity = find_similar_region(input_region)
print("입력된 지역의 키워드: " + final_df[final_df['SIGNGU_NM'] == '고창군']['CORE_KWRD_CN'])

# 출력 형식 설정
print("유사도 Top 5 지역:")
for region, similarity in similar_regions[:5]:
    print(f"{region}: {similarity}")

print()
print(f"입력된 지역 '{input_region}'과(와) 유사도 Top 5 지역 중 가장 낮은 평균 유사도를 가진 지역은 '{lowest_region}', 다른 지역과의 평균 유사도: {lowest_similarity}")
```
